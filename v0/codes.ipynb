# Simulation of Urban MObility (SUMO)

## File structure


| File | Description |
|----------|----------|
| .nod.xml  | It defines the junctions. In our case the start and end point of the road  |
| .edg.xml  | Defines the roads like no of lanes, width, speed limit etc  |
| .rou.xml  | Defines the vehicles and its charcteristics |
| .net.xml  | SUMO road network generated from .nod.xml and .edg.xml|
| .sumocfg  |Tells SUMO which network/route files to load and simulation settings such as timestep and lateral resolution|
| simulation.py|main code|


## .nod.xml


A **Node** represents the starting or ending point of roads, junctions etc.
| Parameter | Description |
|---|---|
| `id` | Unique identifier for each node. |
| `x, y` | Spatial coordinates of the node. |
| `type` | Determines how the traffic is controlled if there are multiple intersections. |

```python

<nodes>
    <node id="n0" x="0" y="0" type="priority"/>
    <node id="n1" x="1000" y="0" type="priority"/>
</nodes>
```

## .edg.xml

An **Edge** represents a road connecting two nodes. It defines the road's direction, number of lanes, speed, and basic lane geometry.

| Parameter | Description |
|---|---|
| `id` | Unique identifier for a road. |
| `from` | ID of the node from where the road starts. |
| `to` | ID of the node where the road ends. |
| `numLanes` | Number of lanes on the road. |
| `speed` | Maximum speed allowed on the road, in m/s. |
| `laneWidth` | Width of each lane in m. |
| `spreadType` | Determines how the lanes are positioned relative to the road's reference line. | 

```python
<edges>
    <edge id="e0"
          from="n0"
          to="n1"
          numLanes="3"
          speed="20.0"
          laneWidth="3.5"
          spreadType="center"/>
</edges>
```

## .rou.xml 

It defines the vehicle charcteristics, the roads the vehicle travels on, how many vehicles enter, when they enter, and their initial conditions.

```python
<routes>

    <vType id="bike" length="1.8" width="0.6"
           guiShape="motorcycle" color="0,1,0"
           accel="3.0" decel="7.0" maxSpeed="35"
           sigma="0.65" minGap="0.383637"
           speedDev="0.45" speedFactor="1.10"
           lcModel="SL2015" lcSublane="8"
           lcPushy="0.85" lcAssertive="0.80" lcImpatience="0.95"
           latAlignment="nice"
           lcStrategic="0" lcCooperative="0.1"
           lcSpeedGain="2.0" lcKeepRight="0"/>

    <route id="r_forward" edges="e0"/>

    <flow id="fwd_bike" type="bike" route="r_forward"
          begin="0" end="3600" vehsPerHour="2160"
          departLane="random" departPos="15" departSpeed="random"/>

</routes>
```

### vType

The `<vType>` defines the physical and driving characteristics of the vehicles.

| Parameter | Description |
|---|---|
| `id` | Unique identifier for the vehicle type. |
| `length` | Length of the vehicle in metres. |
| `width` | Width of the vehicle in metres. |
| `guiShape` | Display style in SUMO-GUI. |
| `color` | Vehicle colour in the GUI. |
| `accel` | Maximum acceleration of the vehicle in m/s². |
| `decel` | Desired deceleration of the vehicle in m/s². |
| `maxSpeed` | Maximum speed capability of the vehicle in m/s (desired speed). |
| `sigma` | Represents driver randomness in vehicle behavior. |
| `minGap` | Desired minimum longitudinal gap between this vehicle and the vehicle ahead, in metres |
| `speedDev` | Standard deviation used when assigning desired speeds to vehicles of this type. |
| `speedFactor` | Multiplier for the desired speed. |

```python
<vType id="bike"
       length="1.8"
       width="0.6"
       guiShape="motorcycle"
       color="0,1,0"

       accel="3.0"
       decel="7.0"
       maxSpeed="35"

       sigma="0.65"
       minGap="0.383637"
       speedDev="0.45"
       speedFactor="1.10"/>
```

| Parameter | Typical/valid range | Meaning |
|---|---:|---|
| `lcModel` | `SL2015` | Selects the SL2015 lane-changing model. |
| `lcSublane` | `0 – 1`* | Controls willingness to change laterally for sublane positioning. |
| `lcPushy` | `0 – 1` | Controls how strongly the vehicle pushes into gaps during lane changes. |
| `lcAssertive` | `0 – 1` | Controls how assertively the vehicle performs lane changes. |
| `lcImpatience` | `0 – 1` | Controls how impatience develops when waiting for a lane change. |
| `latAlignment` | categorical | Specifies the preferred lateral alignment, e.g. `nice`. |
| `lcStrategic` | `0 – 1` | Controls strategic lane-changing motivation. |
| `lcCooperative` | `0 – 1` | Controls willingness to cooperate with other vehicles. |
| `lcSpeedGain` | `0 – 1` | Controls motivation to change lanes for speed gain. |
| `lcKeepRight` | `0 – 1` | Controls motivation to keep right. |

```python
<vType id="bike" length="1.8" width="0.6"
           guiShape="motorcycle" color="0,1,0"
           accel="3.0" decel="7.0" maxSpeed="35"
           sigma="0.65" minGap="0.383637"
           speedDev="0.45" speedFactor="1.10"
           lcModel="SL2015" lcSublane="8"
           lcPushy="0.85" lcAssertive="0.80" lcImpatience="0.95"
           latAlignment="nice"
           lcStrategic="0" lcCooperative="0.1"
           lcSpeedGain="2.0" lcKeepRight="0"/>
```

## .sumocfg

| Section | Parameter | Explanation |
|---|---|---|
| **`input`** | `net-file` | Specifies the **SUMO network file** containing the roads, lanes, junctions, and their geometry. |
| **`input`** | `route-files` | Specifies the **route/traffic file** containing vehicle types, routes, and traffic flows. |
| **`time`** | `begin` | Specifies the **simulation start time**. |
| **`time`** | `end` | Specifies the **simulation end time**. |
| **`time`** | `step-length` | Specifies the **simulation timestep**. SUMO advances the simulation by this amount at each step. |
| **`processing`** | `lateral-resolution` | Specifies the **lateral resolution used by the sublane model**. Smaller values allow finer representation of lateral vehicle movement. |
| **`processing`** | `time-to-teleport` | Controls when SUMO **teleports vehicles that remain stuck for too long**. A value of `-1` disables automatic teleportation. |
| **`report`** | `verbose` | Controls the amount of **detailed output** printed by SUMO. |
| **`report`** | `no-step-log` | Controls whether SUMO prints the regular **simulation-step log** in the terminal. |

```python
<configuration>
    <input>
        <net-file value="straight3lane.net.xml"/>
        <route-files value="straight3lane.rou.xml"/>
    </input>
    <time>
        <begin value="0"/>
        <end value="3600"/>
        <step-length value="0.1"/>
    </time>
    <processing>
        <lateral-resolution value="0.25"/>
        <time-to-teleport value="-1"/>
    </processing>
    <report>
        <verbose value="false"/>
        <no-step-log value="true"/>
    </report>
</configuration>
```

#### TraCI Functions

| TraCI Function | Description |
|---|---|
| `traci.start()` | Starts SUMO and establishes the TraCI connection |
| `traci.simulationStep()` | Advances the SUMO simulation by one simulation step |
| `traci.simulation.getTime()` | Returns the current simulation time in seconds |
| `traci.close()` | Closes the TraCI connection and ends the SUMO session |

| TraCI Function | Description |
|---|---|
| `traci.vehicle.getIDList()` | Returns the IDs of all vehicles currently in the simulation |
| `traci.vehicle.getPosition()` | Returns the vehicle's (x, y) position in meters |
| `traci.vehicle.getSpeed()` | Returns the vehicle's speed in m/s |
| `traci.vehicle.getAcceleration()` | Returns the vehicle's acceleration in m/s² |
| `traci.vehicle.getAngle()` | Returns the vehicle's heading angle in degrees |
| `traci.vehicle.getLength()` | Returns the vehicle's length in meters |
| `traci.vehicle.getWidth()` | Returns the vehicle's width in meters |


| TraCI Function | Description |
|---|---|
| `traci.vehicle.setAcceleration()` | Commands a vehicle to accelerate/decelerate at a specified acceleration for a specified duration |
| `traci.vehicle.setSpeed()` | Sets the vehicle's desired speed |
| `traci.vehicle.setSpeedMode()` | Controls which SUMO speed control behaviour|
| `traci.vehicle.setLaneChangeMode()` | Controls SUMO's automatic lane changing behavior |

| TraCI Function | Description |
|---|---|
| `traci.vehicle.moveToXY()` | Moves a vehicle to a specified Cartesian (x, y) position |
| `traci.vehicle.subscribe()` | Subscribes to vehicle variables so their values can be retrieved automatically each simulation step |
| `traci.vehicle.getSubscriptionResults()` | Retrieves the latest values for the variables previously subscribed to |
